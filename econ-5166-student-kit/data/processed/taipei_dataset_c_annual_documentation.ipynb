### Missingness、coverage與低樣本品質

```python
rows=[]
for key,g in frame.groupby(groups):
    vals=key if isinstance(key,tuple) else (key,)
    info=dict(zip(groups,vals))
    for t in u.THRESHOLDS:rows.append(dict(info,threshold=t,count_below=int(g.observed_accident_count.lt(t).sum()),zero_observed=int(g.observed_accident_count.eq(0).sum())))
u.show('五個低樣本候選門檻；未排除任何里',pd.DataFrame(rows))
u.show('各傷亡欄全缺失／部分缺失旗標',frame.groupby(groups)[[s+suffix for s in u.CASUALTIES for suffix in ['_all_missing_flag','_partial_missing_flag']]].sum().reset_index())
u.show('公館里保留',frame.loc[frame.village_id.eq('6301100044')].iloc[:,:18])
u.show('糖廍里同規則保留',frame.loc[frame.standard_village_name.eq('糖廍里')].iloc[:,:18])
```
