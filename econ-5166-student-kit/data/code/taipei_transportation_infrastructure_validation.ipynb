### 補充原欄品質

道路個別線長、原ID／分類、號誌區名及入口複合標籤鍵；不因屬性值相同就刪除實體紀錄。

```python
import numpy as np
roads=ctx.layers['osm']
show('OSM研究區bbox線長分布（EPSG3826公尺，非VKT）',roads.length.describe(percentiles=[.01,.25,.5,.75,.95,.99]).rename('length_m').reset_index())
show('OSM識別碼與分類基本品質',pd.DataFrame([dict(loaded_lines=len(roads),missing_osm_id=int(roads.osm_id.isna().sum()),repeated_osm_id_rows=int(roads.osm_id.duplicated(keep=False).sum()),missing_fclass=int(roads.fclass.isna().sum()),unclassified_source_codes=int(roads['_research_road_group'].eq('unclassified_source_code').sum()),nonpositive_line_length=int(roads.length.le(0).sum()))]))
sig=ctx.layers['signals'];sig=sig[sig['_geometry_eligible']].copy()
pairs=ctx.btree.query(np.asarray(sig.geometry.array,dtype=object),predicate='within');occ=np.bincount(pairs[0],minlength=len(sig));assigned=np.full(len(sig),-1)
for i,j in zip(*pairs):
 if occ[i]==1:assigned[i]=j
sig['diagnostic_spatial_district']=[ctx.districts[j] if j>=0 else '' for j in assigned]
sig['district_label_conflict']=(assigned>=0)&sig['行政區'].ne(sig['diagnostic_spatial_district'])
show('號誌來源行政區與唯一空間歸區衝突（保留原值）',sig.loc[sig.district_label_conflict,['流水號','地點','行政區','diagnostic_spatial_district','WGS經度座標','WGS緯度座標']])
ctx.summaries['signals_district_conflicts']=int(sig.district_label_conflict.sum())
mrt=ctx.layers['mrt'];key=['出入口名稱','出入口編號']
show('捷運入口複合標籤鍵重複診斷（非正式站碼）',mrt.loc[mrt.duplicated(key,keep=False),['項次',*key,'經度','緯度']])
ctx.summaries['mrt_composite_label_duplicate_rows']=int(mrt.duplicated(key,keep=False).sum())
```

### 號誌觀察單位補充限制

```python
signal_names=ctx.signal_points['地點'].fillna('')
testing=signal_names.str.contains('測試',regex=False)
label_review=signal_names.str.contains('測試|LCS',case=False,regex=True)
show('名稱明示測試或LCS之待審單位；不全當實體路口',ctx.signal_points.loc[label_review,['流水號','地點','行政區','WGS經度座標','WGS緯度座標']])
ctx.summaries['signals_named_test_count']=int(testing.sum())
ctx.summaries['signals_named_test_or_LCS_count']=int(label_review.sum())
ctx.summaries['signals_named_test_or_LCS_records']=ctx.signal_points.loc[label_review,['流水號','地點','行政區','WGS經度座標','WGS緯度座標']].to_dict('records')
condition='118筆區名不一致及名稱含測試／LCS等單位待審；本表只計觀測位置紀錄。正式實體路口需先確認排除／合併規則，不能把全部列當路口。'
mask=ctx.feature_table.feature.str.startswith('signalized_')
ctx.feature_table.loc[mask,'limitation']=ctx.feature_table.loc[mask,'limitation']+'；'+condition
ctx.readiness_table.loc[ctx.readiness_table.component.eq('B3 signals'),'conditions']+='；'+condition
show('號誌候選指標之補充條件',ctx.feature_table[mask],block=3)
show('號誌可用性補充：Ready僅指有明示條件的位置紀錄，非無條件實體路口',ctx.readiness_table[ctx.readiness_table.component.eq('B3 signals')])
print('名稱含測試',int(testing.sum()),'名稱含測試或LCS',int(label_review.sum()),'；本單元保留原始記錄，未採正式排除政策。')
```
