## 1. Accident CRS diagnostics — definitions, operations and limitations
原始X解讀為longitude、Y為latitude；CRS官方軸序與程式輸入軸序分開，所有Transformer使用always_xy=True。只比較EPSG:4326與3824至3826。以本機PROJ／EPSG資料庫列正式CRS定義、datum、axis、operation、accuracy、grid及ballpark，不以配對率認證原datum。

技術定義參照[PyProj Transformer文件](https://pyproj4.github.io/pyproj/stable/api/transformer.html)：accuracy=-1表示未知，always_xy維持經度／緯度或東距／北距。此為軟體行為依據，不是事故資料來源的datum證明。


```python
import pyproj
from pyproj import CRS,database
from pyproj.transformer import TransformerGroup,AreaOfInterest
from pyproj import network
network.set_network_enabled(False)
print({'pyproj':pyproj.__version__,'PROJ':pyproj.proj_version_str,'EPSG_version':database.get_database_metadata('EPSG.VERSION'),'EPSG_date':database.get_database_metadata('EPSG.DATE'),'network_enabled':network.is_network_enabled()})
for epsg in [4326,3824,3826]:
    c=CRS.from_epsg(epsg)
    print(json.dumps(dict(epsg=epsg,name=c.name,datum=c.datum.name,axes=[dict(name=a.name,abbrev=a.abbrev,direction=a.direction,unit=a.unit_name) for a in c.axis_info],definition=c.to_wkt()),ensure_ascii=False))
transforms={};operation_rows=[];grid_rows=[]
for epsg in [4326,3824]:
    group=TransformerGroup(CRS.from_epsg(epsg),CRS.from_epsg(3826),always_xy=True,area_of_interest=AreaOfInterest(121.3,24.8,121.8,25.3),allow_ballpark=True)
    assert group.transformers
    t=group.transformers[0];transforms[epsg]=t
    operation_rows.append(dict(source_epsg=epsg,best_available=group.best_available,available_operations=len(group.transformers),unavailable_operations=len(group.unavailable_operations),selected_operation=t.description,accuracy_m=t.accuracy,pipeline=t.definition,ballpark=any(o.has_ballpark_transformation for o in t.operations),required_grids=sum(len(o.grids) for o in t.operations)))
    print('EPSG',epsg,'operation details',t.to_json_dict())
    for o in list(t.operations)+list(group.unavailable_operations):
        for g in o.grids:grid_rows.append(dict(source_epsg=epsg,operation=o.name,grid=g.short_name,available=g.available,url=g.url))
display(pd.DataFrame(operation_rows));show('所需／缺少Grid（零列表示此operation不要求grid，非已驗證測量精度）',pd.DataFrame(grid_rows))
projected={};differences=[]
for year,f in candidates.items():
    lon=pd.to_numeric(f['座標-X'],errors='coerce').to_numpy(float);lat=pd.to_numeric(f['座標-Y'],errors='coerce').to_numpy(float)
    parseable=np.isfinite(lon)&np.isfinite(lat)
    degree_ok=parseable&(lon>=-180)&(lon<=180)&(lat>=-90)&(lat<=90)
    for epsg,t in transforms.items():
        x=np.full(len(f),np.nan);y=x.copy();x[degree_ok],y[degree_ok]=t.transform(lon[degree_ok],lat[degree_ok],errcheck=True)
        projected[(year,epsg)]={'x':x,'y':y,'parseable':parseable,'degree_ok':degree_ok}
    a=projected[(year,4326)];b=projected[(year,3824)];d=np.hypot(a['x']-b['x'],a['y']-b['y'])
    differences.append(dict(year=year,N=len(f),parseable=int(parseable.sum()),degree_valid=int(degree_ok.sum()),min_m=float(np.nanmin(d)),q25_m=float(np.nanquantile(d,.25)),median_m=float(np.nanmedian(d)),q75_m=float(np.nanquantile(d,.75)),q95_m=float(np.nanquantile(d,.95)),q99_m=float(np.nanquantile(d,.99)),max_m=float(np.nanmax(d))))
diff_summary=pd.DataFrame(differences);display(diff_summary)
print('比較的是目前PROJ所選operation的數值結果；候選輸出即使相同，也不代表datum相同或來源datum已確認。TWD97與WGS84近似／等價實作可能使本診斷無辨識力。')
```

## 2. Real accident within / covered_by audit
保留456個原幾何。STRtree對每候選點分別query within及covered_by；不取第一筆、不用最近里。互斥狀態：unique_clear、shared_boundary、shared_vertex、positive_area_overlap、other_multiple、outside_or_hole、missing_or_unparseable；若發現單一外邊界候選另列single_boundary_pending，度數異常另列invalid_coordinate_range。共同頂點採原始vertex精確相等，不任意設snap容差。

「多里內部within」為正面積重疊；全在邊界且至少兩候選，再以共同vertex判斷。within一筆但covered_by多筆不算明確唯一。界外與內部孔洞分開作輔助標記，原始最終狀態仍保留outside_or_hole。


```python
bgeom=np.asarray(boundary.geometry.array,dtype=object);btree=s.STRtree(bgeom)
def rings(g):
    if g.geom_type=='Polygon':return [g.exterior,*g.interiors]
    return [r for p in g.geoms for r in [p.exterior,*p.interiors]]
vertex_sets=[{tuple(c[:2]) for ring in rings(g) for c in ring.coords} for g in bgeom]
holes=[s.Polygon(r) for g in bgeom for ring in ([g] if g.geom_type=='Polygon' else list(g.geoms)) for r in ring.interiors]
hole_union=s.union_all(holes)
audits={};join_summary=[]
for (year,epsg),p in projected.items():
    f=candidates[year];valid=np.flatnonzero(p['degree_ok']);points=s.points(p['x'][valid],p['y'][valid])
    within=[[] for _ in range(len(f))];covered=[[] for _ in range(len(f))]
    for predicate,container in [('within',within),('covered_by',covered)]:
        q=btree.query(points,predicate=predicate)
        for local,j in zip(*q):container[valid[local]].append(int(j))
    rows=[]
    record_ids=f.record_id.to_numpy();source_paths=f.source_file.to_numpy();source_rows=f.source_row_number.to_numpy()
    for i in range(len(f)):
        w=sorted(within[i]);c=sorted(covered[i]);subtype=''
        if not p['parseable'][i]:status='missing_or_unparseable'
        elif not p['degree_ok'][i]:status='invalid_coordinate_range'
        elif len(w)==len(c)==1 and w==c:status='unique_clear'
        elif len(w)>1:status='positive_area_overlap'
        elif len(c)>=2 and not w:
            xy=(p['x'][i],p['y'][i]);status='shared_vertex' if sum(xy in vertex_sets[j] for j in c)>=2 else 'shared_boundary'
        elif len(c)>1 or len(w)>1:status='other_multiple'
        elif len(c)==1 and len(w)==0:status='single_boundary_pending'
        elif not c and not w:
            status='outside_or_hole';pt=s.Point(p['x'][i],p['y'][i]);subtype='inside_original_hole' if hole_union.covers(pt) else 'outside_fixed_union'
        else:status='other_multiple'
        rows.append(dict(record_id=record_ids[i],source_file=source_paths[i],source_row_number=int(source_rows[i]),year=year,crs_candidate=epsg,within_count=len(w),covered_by_count=len(c),within_villages=';'.join(village_ids[w]),covered_villages=';'.join(village_ids[c]),candidate_districts=';'.join(sorted(set(boundary.iloc[c].TNAME))),status=status,outside_subtype=subtype,clear_village_id=village_ids[c[0]] if status=='unique_clear' else '',x_3826=p['x'][i],y_3826=p['y'][i]))
    audit=pd.DataFrame(rows)
    assert len(audit)==len(f) and audit.record_id.is_unique and set(audit.record_id)==set(f.record_id)
    assert audit.status.notna().all() and audit.groupby('status').size().sum()==len(f)
    audits[(year,epsg)]=audit
    for status in ['unique_clear','shared_boundary','shared_vertex','positive_area_overlap','other_multiple','outside_or_hole','missing_or_unparseable','single_boundary_pending','invalid_coordinate_range']:
        status_n=int(audit.status.eq(status).sum())
        join_summary.append(dict(year=year,crs_candidate=epsg,status=status,N=status_n,ratio_all=status_n/len(f),ratio_parseable=status_n/p['parseable'].sum() if status!='missing_or_unparseable' else np.nan,all_candidates=len(f),parseable=int(p['parseable'].sum())))
join_summary=pd.DataFrame(join_summary);show('逐年度、逐候選CRS互斥配對狀態；分母明列',join_summary)
show('樣本稽核表前十列（全45130列各候選版本均保留於記憶體並assert完整）',audits[(2024,4326)].head(10))
crs_changes=[]
for year in [2024,2025]:
    a=audits[(year,4326)].set_index('record_id');b=audits[(year,3824)].set_index('record_id').reindex(a.index)
    change=a.status.ne(b.status)|a.within_villages.ne(b.within_villages)|a.covered_villages.ne(b.covered_villages)
    crs_changes.append(dict(year=year,status_changed=int(a.status.ne(b.status).sum()),within_candidates_changed=int(a.within_villages.ne(b.within_villages).sum()),covered_candidates_changed=int(a.covered_villages.ne(b.covered_villages).sum()),any_status_or_village_changed=int(change.sum())))
display(pd.DataFrame(crs_changes))
print('未輸出正式事故歸里資料；配對率與候選一致不構成官方datum認證。')
```

```python
cats=sorted(set(category)-{'non_zoning_boundary','unclassified'})
unions={c:s.union_all(reference[category==c]) for c in cats}
classified=s.union_all(list(unions.values()));unknown_union=s.union_all(reference[unknown]);candidate=s.union_all([classified,unknown_union])
print('City reference class unions complete; separating area and non-area contacts.')
cross_parts=[];contact_audit=[];contact_geometries={}
for i,c in enumerate(cats):
    for d in cats[i+1:]:
        x=unions[c].intersection(unions[d]);pp,rem=parts(x)
        if rem:contact_geometries[(c,d)]=rem
        contact_audit.append(dict(category_a=c,category_b=d,positive_area_m2=x.area,nonarea_parts=len(rem),nonarea_length_m=sum(g.length for g in rem),nonarea_types=','.join(sorted(set(g.geom_type for g in rem)))))
        if pp:cross_parts.append(s.union_all(pp))
unknown_intersection=classified.intersection(unknown_union)
unknown_parts,unknown_contacts=parts(unknown_intersection)
contact_geometries[('classified','unknown')]=unknown_contacts
contact_audit.append(dict(category_a='classified',category_b='unknown',positive_area_m2=unknown_intersection.area,nonarea_parts=len(unknown_contacts),nonarea_length_m=sum(g.length for g in unknown_contacts),nonarea_types=','.join(sorted(set(g.geom_type for g in unknown_contacts)))))
cross_unknown=s.union_all(unknown_parts)
unresolved_overlap=s.union_all([*cross_parts,cross_unknown])
show('跨類及未知交集的非面接觸稽核：線點完整留記憶體，只面構件入面積扣除',pd.DataFrame(contact_audit))
assert abs(unresolved_overlap.area-s.union_all([s.union_all(cross_parts),cross_unknown]).area)<1e-6
print('Positive-area overlap masks complete; zero-area contact geometries retained separately.')
withheld=s.union_all([unresolved_overlap,geometry_envelope,class_review_area])
print('Constructing conservative effective area masks.')
effective=classified.difference(withheld)
effective_unions={c:g.difference(withheld) for c,g in unions.items()}
uncovered=city.difference(candidate)
uncertainty=s.union_all([unknown_union,uncovered,unresolved_overlap,geometry_envelope,class_review_area]).intersection(city)
main_cats=['residential','commercial','industrial','educational','park_green']
effective_main=s.union_all([effective_unions[c] for c in main_cats]);effective_other=effective.difference(effective_main)
coverage=base.copy()
GEOMS={'candidate_zoning_coverage':candidate,'classified_zoning_coverage':classified,'effective_classified_area':effective,'effective_main_five_area':effective_main,'effective_other_classified_area':effective_other,'unclassified_area':unknown_union,'unresolved_overlap_area':unresolved_overlap,'uncovered_area':uncovered,'unresolved_geometry_area':geometry_envelope,'classification_review_area':class_review_area,'total_uncertainty_area':uncertainty}
def village_areas(g):
    # Associativity: V is contained in its original district union, so
    # Area(V intersect G)=Area(V intersect (G intersect district_union)).
    # This reduces repeated overlay with the entire city's complex geometry;
    # original village geometries and denominators are unchanged.
    values=np.zeros(len(boundary),dtype=float)
    for district,dshape in district_shapes.items():
        ids=np.flatnonzero(boundary.TNAME.eq(district).to_numpy())
        clipped=g.intersection(dshape)
        values[ids]=s.area(s.intersection(bgeom[ids],clipped))
    return values
for name,g in GEOMS.items():
    print('Village diagnostics:',name)
    coverage[name]=village_areas(g)
check_ids=np.array([0,100,200,300,455])
for name in ['candidate_zoning_coverage','effective_classified_area','total_uncertainty_area']:
    np.testing.assert_allclose(coverage.loc[check_ids,name].to_numpy(),s.area(s.intersection(bgeom[check_ids],GEOMS[name])),rtol=1e-10,atol=.001)
print('District-union partition optimization agrees with direct whole-city intersections on fixed check villages; original village geometry unchanged.')
for name,numerator in [('candidate_coverage_ratio','candidate_zoning_coverage'),('classified_coverage_ratio','classified_zoning_coverage'),('effective_coverage_ratio','effective_classified_area'),('unclassified_ratio','unclassified_area'),('unresolved_overlap_ratio','unresolved_overlap_area'),('uncovered_ratio','uncovered_area'),('unresolved_geometry_ratio','unresolved_geometry_area'),('classification_review_ratio','classification_review_area'),('uncertainty_ratio','total_uncertainty_area')]:
    coverage[name]=coverage[numerator]/coverage.village_area_m2
    assert coverage[name].between(-1e-9,1+1e-9).all(),name
np.testing.assert_allclose(coverage.candidate_zoning_coverage+coverage.uncovered_area,coverage.village_area_m2,atol=.01)
np.testing.assert_allclose(coverage.effective_classified_area+coverage.total_uncertainty_area,coverage.village_area_m2,atol=.01)
city_summary=pd.DataFrame([dict(set=name,city_union_area_km2=g.intersection(city).area/1e6) for name,g in GEOMS.items()])
display(city_summary)
unknown_source=urban.loc[unknown,['編號']+KEYS].copy();unknown_source['reference_total_km2']=s.area(reference[unknown])/1e6;unknown_source['inside_city_km2']=[g.intersection(city).area/1e6 for g in reference[unknown]]
show('47.93km²待確認面積的原始來源',unknown_source)
show('456里全部候選面積及覆蓋診斷（非正式B2）',coverage,block=16)
display(coverage[['candidate_coverage_ratio','effective_coverage_ratio','unclassified_ratio','uncovered_ratio','uncertainty_ratio']].describe(percentiles=[.01,.05,.25,.5,.75,.95,.99]).T)
thresholds=pd.DataFrame([dict(threshold=t,candidate_villages=int(coverage.candidate_coverage_ratio.ge(t).sum()),classified_villages=int(coverage.classified_coverage_ratio.ge(t).sum()),effective_villages=int(coverage.effective_coverage_ratio.ge(t).sum()),effective_village_year_rows=2*int(coverage.effective_coverage_ratio.ge(t).sum())) for t in [.5,.7,.8,.9,.95,.99]])
display(thresholds)
low=coverage[coverage.effective_coverage_ratio.lt(.9)].sort_values('effective_coverage_ratio')
show('低於90%有效覆蓋之完整診斷清單；90%只是展示門檻，非正式排除規則',low[['village_id','district_name','raw_village_name','effective_coverage_ratio','candidate_coverage_ratio','unclassified_ratio','unresolved_geometry_ratio','uncovered_ratio']])
district_coverage=coverage.groupby('district_name').agg(villages=('village_id','size'),area_m2=('village_area_m2','sum'),candidate_m2=('candidate_zoning_coverage','sum'),effective_m2=('effective_classified_area','sum'),unknown_m2=('unclassified_area','sum'),uncovered_m2=('uncovered_area','sum'),geometry_pending_m2=('unresolved_geometry_area','sum'),below90=('effective_coverage_ratio',lambda x:x.lt(.9).sum()))
for name,numerator in [('candidate_ratio','candidate_m2'),('effective_ratio','effective_m2'),('unknown_ratio','unknown_m2'),('uncovered_ratio','uncovered_m2')]:district_coverage[name]=district_coverage[numerator]/district_coverage.area_m2
display(district_coverage)
print('行政區表為固定里面積加總加權，保留原里界微量重疊限制；全市總量另用city union，不冒稱兩者完全等同。山區集中僅依位置／行政區描述，沒有高程資料，未正式辨識地形因果。')
```

## 7. B2 zero, unknown and conditional interval rules
五主要變數共同分母為原里面積。下界L_c=有效類別面∩里／里面積；若未知、未覆蓋、修復／分類待審或重疊存在，它是**已辨識覆蓋比例／條件性保守下界**，不是完整行政里分區比例。

在「已接受面／分類正確、未確認範圍全部包含在本次uncertainty集合、完整圖資可在其中補足」假設下，可列U_c=min(1,L_c+uncertainty_ratio)。這是資料情境範圍，不是信賴區間；因真實修復或缺圖邊界可能超出三方法包絡，不能當作官方真實分區比例的無條件上界。若不接受此包絡假設，完整比例上下界標為不可認證，只交付觀測L與品質旗標。

四狀態：完整可計算正值、完整已知零、部分觀測正下界、無充分資料估完整比例。只有無未知且有效交集為零才可稱指定分區已知不存在。10⁻⁶m²僅浮點等價容差，不代表忽略有意義的缺圖。

A品質充足子樣本：50–99%門檻逐一比較，不能把門檻內殘餘未知變成已知；B保留456里觀測下界與品質旗標，但環境分群會受缺圖影響；C各門檻／包絡情境作穩健性，不依分群顯著性挑門檻；D主要D原版先採B1，B2另列補充。最終分群樣本與最低品質門檻仍待審。


```python
feature_names={c:c+'_zoning_ratio' for c in main_cats}
b2=coverage[['village_id','district_name','raw_village_name','effective_coverage_ratio','uncertainty_ratio']].copy();b2_states=[]
for c,name in feature_names.items():
    a=village_areas(effective_unions[c]);L=a/coverage.village_area_m2.to_numpy();U=np.minimum(1,L+coverage.uncertainty_ratio.to_numpy())
    b2[name+'_observed_lower']=L;b2[name+'_conditional_upper']=U
    known=coverage.total_uncertainty_area.le(1e-6).to_numpy();positive=a>1e-6
    state=np.where(known,np.where(positive,'known_computable_complete','known_zero'),np.where(positive,'partial_observed_lower_bound','insufficient_for_complete_ratio'))
    b2[name+'_state']=state
    for st,n in pd.Series(state).value_counts().items():b2_states.append(dict(variable=name,status=st,villages=int(n)))
    assert np.all(L>=-1e-9) and np.all(U<=1+1e-9) and np.all(U+1e-9>=L)
show('五主候選的零值／未知狀態分布',pd.DataFrame(b2_states))
show('條件性區間及品質旗標前十里；不是正式B2',b2.head(10))
print('不確定集合不包含其他已知類別：特殊／混合及非主類有效面積仍是已知，不一律設未知。未由B2覆蓋不足刪除事故或人口里。')
```
