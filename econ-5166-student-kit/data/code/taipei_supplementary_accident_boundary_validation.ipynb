### 保留傷亡差值及非唯一缺座標候選

逐筆保存原死亡／2–30日／受傷與全國值。3筆缺座標即使無唯一事件對照，仍列出同時間／地址的全部位置候選及來源，避免將「沒有唯一補值證據」誤稱補充來源沒有座標；不修改原值。

表中制式「未形成唯一事件對照」狀態只表示未取得可直接採用的唯一座標補值證據；其中2025太原路50號已有A級事件配對，但兩份都缺座標，並非撤銷其A級事件對照。其他同鍵多重仍不強派。

```python
import numpy as np
casualty_differences=[];ambiguous_positions=[]
for year,local in ctx.candidates().items():
    aligned=ctx.cache['national_accidents_aligned'];other=aligned['other'][year];matches=aligned['matches'][year]
    supported=matches[matches.match_grade.str.startswith(('A_','C_'))]
    a=local.iloc[supported.local_index.to_numpy()].reset_index(drop=True);b=other.iloc[supported.other_index.to_numpy()].reset_index(drop=True)
    counts=b['死亡受傷人數'].str.extract(r'^死亡(\d+);受傷(\d+)$').apply(pd.to_numeric,errors='coerce')
    death=pd.to_numeric(a['死亡人數']);late=pd.to_numeric(a['2-30日死亡人數']);injury=pd.to_numeric(a['受傷人數'])
    affected=death.ne(counts[0])|injury.ne(counts[1])|late.gt(0)
    for i in np.flatnonzero(affected.to_numpy()):
        casualty_differences.append(dict(year=year,record_id=a.iloc[i].record_id,local_address=a.iloc[i]['_address'],national_address=b.iloc[i]['發生地點'],local_death=a.iloc[i]['死亡人數'],local_2_30=a.iloc[i]['2-30日死亡人數'],local_injuries=a.iloc[i]['受傷人數'],national_casualties=b.iloc[i]['死亡受傷人數'],national_member=b.iloc[i]['_source_member'],national_row=int(b.iloc[i]['_source_row']),match_grade=supported.iloc[i].match_grade,interpretation='可能時間口徑／公開版本或候選配對差異，未斷定錯誤或覆寫'))
    missing=local[~np.isfinite(local['_lon'])|~np.isfinite(local['_lat'])]
    for source in ['accident_points','national_accidents']:
        r=ctx.cache[source+'_aligned']['other'][year]
        for _,row in missing.iterrows():
            time=r['_minute'].eq(row['_minute']);exact=time&r['_address_valid']&r['_address'].eq(row['_address'])
            canonical=ctx.canonical_address(row['_address'])
            normalized=time&r['_address_valid']&r['_address'].map(lambda x:canonical is not None and ctx.canonical_address(x)==canonical)
            selection=exact if exact.any() else normalized if normalized.any() else time
            selected=r[selection];distinct=selected[['_lon','_lat']].dropna().drop_duplicates().shape[0]
            if not len(selected):
                ambiguous_positions.append(dict(year=year,record_id=row.record_id,source=source,candidate_N=0,valid_distinct_coordinates=0,status='無可核對候選；不補值'))
            for _,candidate in selected.iterrows():
                path=candidate['_source_path'];valid=bool(np.isfinite(candidate['_lon']) and np.isfinite(candidate['_lat']))
                ambiguous_positions.append(dict(year=year,record_id=row.record_id,local_address=row['_address'],source=source,candidate_N=len(selected),valid_distinct_coordinates=distinct,matching_evidence='原文地址與分鐘相同但未唯一確認' if exact.any() else '格式地址與分鐘候選' if normalized.any() else '只有時間相同，地址未證',candidate_address=safe_text(candidate['_address']),candidate_lon=candidate['_lon'],candidate_lat=candidate['_lat'],valid_numeric_coordinates=valid,source_path=path,source_member=candidate.get('_source_member',''),source_row=int(candidate['_source_row']),source_sha256=next(x['sha256'] for x in ctx.inventory if x['path']==path),status='未形成唯一事件對照；非正式補值、未套用'))
show('傷亡值不同或涉及2–30日死亡的完整候選核對；保留兩來源值',pd.DataFrame(casualty_differences),block=8)
show('3筆缺座標的所有同時間／地址候選位置，保留多重與無效候選；不填補',pd.DataFrame(ambiguous_positions),block=6)
ctx.summaries['casualty_difference_records']=casualty_differences
ctx.summaries['missing_coordinate_nonunique_candidates']=ambiguous_positions
```

### 異常群組完整分母與候選替代位置

完整保留869＋2＋44＋3＝918筆的分母，未配對者也列入。替代位置只在A／C地址支持對應下試算，不直接採用；B純時間候選不升格位置認證。

```python
import numpy as np
groups=[];position_evidence=[]
for source in ['accident_points','national_accidents']:
    aligned=ctx.cache[source+'_aligned']
    for year in [2024,2025]:
        old=ctx.audits[(year,'fixed')];match=aligned['matches'][year]
        conditions=[old.district_conflict,old.status.eq('unique_clear')&~old.district_conflict&old.address_label_conflict,old.status.eq('outside_or_hole'),old.status.eq('missing_or_unparseable')]
        labels=['原區名衝突','額外地址衝突','界外或孔洞','缺座標']
        category=np.select(conditions,labels,default='其他')
        frame=match.copy();frame['original_group']=category
        frame=frame[frame.original_group.ne('其他')]
        tab=frame.groupby(['original_group','match_grade']).size().rename('N').reset_index()
        tab['source']=source;tab['year']=year;groups.append(tab)
        alternative=ctx.assign(aligned['other'][year])
        for label in labels:
            subset=frame[frame.original_group.eq(label)];supported=subset[subset.match_grade.str.startswith(('A_','C_'))]
            detail=aligned['details'];detail=detail[detail.year.eq(year)&detail.local_index.isin(subset.local_index)]
            agree=changed=unique=0
            for row in supported.itertuples(index=False):
                a=old.iloc[row.local_index];b=alternative.iloc[row.other_index]
                unique+=int(b.status=='unique_clear')
                agree+=int(b.status=='unique_clear' and b.spatial_district==a.raw_district)
                changed+=int(b.village_id!=a.village_id or b.status!=a.status)
            position_evidence.append(dict(source=source,year=year,original_group=label,original_N=len(subset),address_supported_A_C=len(supported),time_only_B=int(subset.match_grade.str.startswith('B_').sum()),unmatched=int(subset.other_index.lt(0).sum()),matched_coordinate_same=int(detail.coordinate_class.eq('A_same').sum()),matched_one_missing=int(detail.coordinate_class.eq('B_one_missing').sum()),matched_both_missing=int(detail.coordinate_class.eq('both_missing').sum()),matched_difference_gt1m=int(detail.difference_m.gt(1).sum()),A_C_alternative_unique=unique,A_C_alternative_agrees_original_district=agree,A_C_village_or_status_changed=changed,policy='僅候選位置診斷，不套用更正／補值'))
result=pd.concat(groups,ignore_index=True)
assert result.groupby('source').N.sum().eq(918).all()
show('異常群組完整分母與A／C／B／E等級，含未配對者',result)
show('替代位置對既有異常的支持與限制；C仍非官方位置認證',pd.DataFrame(position_evidence))
ctx.summaries['anomaly_group_review']=result.to_dict('records')
ctx.summaries['anomaly_position_review']=position_evidence
```
